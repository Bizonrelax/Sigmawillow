```python
"""
T_compute.py

Вычисляет T(K) — число пар двузначных чисел (A, B) с A < B
и перекрёстной разностью yu − xv = K.

Также вычисляет |S(K)| — число всех упорядоченных пар (A, B)
с Δ(A, B) = K, без фильтра A < B.

Проверяет:
  1. Закон сохранения: сумма T(K) по всем K равна 4005.
  2. Симметрию: T(K) + T(−K) = |S(K)| для K ≠ 0.
  3. Гипотезу об асимметрии: T(K) ≠ T(−K) для 1 ≤ |K| ≤ 81.

Выводит:
  - отчёт о проверках в консоль;
  - таблицу T(K) в Markdown (файл T_table.md);
  - таблицу T(K) в CSV (файл T_table.csv).

Запуск:
  python3 T_compute.py

Автор не указан. Лицензия: MIT.
"""

from collections import defaultdict


def cross_difference(a, b):
    """Возвращает yu − xv для A = 10x + y, B = 10u + v."""
    x, y = divmod(a, 10)
    u, v = divmod(b, 10)
    return y * u - x * v


def compute_tables(k_min=-500, k_max=500):
    """
    Возвращает два словаря:
      T[K] — число пар (A, B) с 10 ≤ A < B ≤ 99 и Δ(A, B) = K;
      S[K] — число пар (A, B) с 10 ≤ A, B ≤ 99 и Δ(A, B) = K.
    """
    T = defaultdict(int)
    S = defaultdict(int)
    for a in range(10, 100):
        for b in range(10, 100):
            k = cross_difference(a, b)
            if k_min <= k <= k_max:
                S[k] += 1
                if a < b:
                    T[k] += 1
    return T, S


def check_conservation(T):
    """Сумма T(K) по всем K должна равняться 4005 = C(90, 2)."""
    total = sum(T.values())
    expected = 90 * 89 // 2
    print("Проверка закона сохранения:")
    print(f"  Сумма T(K) = {total}")
    print(f"  Ожидалось   = {expected}")
    print(f"  Совпадает   = {total == expected}")
    print()
    return total == expected


def check_symmetry(T, S):
    """T(K) + T(−K) = |S(K)| для K ≠ 0."""
    print("Проверка симметрии T(K) + T(−K) = |S(K)|:")
    ok = True
    for k in range(1, 100):
        lhs = T.get(k, 0) + T.get(-k, 0)
        rhs = S.get(k, 0)
        if lhs != rhs:
            print(f"  НАРУШЕНИЕ при K = {k}: {lhs} != {rhs}")
            ok = False
    print(f"  Все проверки пройдены: {ok}")
    print()
    return ok


def check_asymmetry_hypothesis(T):
    """
    Гипотеза: T(K) ≠ T(−K) для всех K с 1 ≤ |K| ≤ 81.
    Проверяем полным перебором.
    """
    print("Проверка гипотезы об асимметрии T(K) ≠ T(−K) для 1 ≤ |K| ≤ 81:")
    violations = []
    for k in range(1, 82):
        if T.get(k, 0) == T.get(-k, 0):
            violations.append(k)
    if violations:
        print(f"  Гипотеза ОПРОВЕРГНУТА при K = {violations}")
    else:
        print("  Гипотеза ПОДТВЕРЖДЕНА для всех 1 ≤ |K| ≤ 81")
    print()
    return violations


def write_markdown(T, S, path="T_table.md"):
    """Сохраняет таблицу в Markdown."""
    with open(path, "w", encoding="utf-8") as f:
        f.write("# Таблица T(K) для перекрёстной разности\n\n")
        f.write("T(K) — число пар (A, B) двузначных чисел с A < B и Δ(A, B) = K.\n\n")
        f.write("| K | T(K) | S(K) | T(K) + T(−K) |\n")
        f.write("|---|------|------|--------------|\n")
        for k in range(-81, 82):
            sym = T.get(k, 0) + T.get(-k, 0)
            f.write(f"| {k} | {T.get(k, 0)} | {S.get(k, 0)} | {sym} |\n")
    print(f"Таблица сохранена в {path}")


def write_csv(T, S, path="T_table.csv"):
    """Сохраняет таблицу в CSV."""
    with open(path, "w", encoding="utf-8") as f:
        f.write("K,T(K),S(K),T(K)+T(-K)\n")
        for k in range(-81, 82):
            sym = T.get(k, 0) + T.get(-k, 0)
            f.write(f"{k},{T.get(k, 0)},{S.get(k, 0)},{sym}\n")
    print(f"Таблица сохранена в {path}")


def main():
    T, S = compute_tables(k_min=-500, k_max=500)
    check_conservation(T)
    check_symmetry(T, S)
    check_asymmetry_hypothesis(T)
    write_markdown(T, S)
    write_csv(T, S)


if __name__ == "__main__":
    main()
```
